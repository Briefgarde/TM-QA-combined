This script fixes a small problem that occured with 'pubmedqa_oracle_pool.json'. All of the entry in this file got duplicated during building, so the file contain each entry twice. This script simply filter the file to only keep one of each. 

```python
import sys
sys.path.append('../')
import json
from pprint import pp
```

```python
oracle_path = "evaluation/pubmedqa_oracle_pool.json"
with open(oracle_path, "r", encoding="utf-8") as oracle_pool_json:
    oracle_pool = json.load(oracle_pool_json)

pp(oracle_pool[0])
```

```python
print(f"length of base pool : {len(oracle_pool)}")

set_id = set()

for entry in oracle_pool:
    id = entry["id"]
    set_id.add(id)
print(f"length of unique id in pool : {len(set_id)}")
```

```python
dup = set()
oracle_pool_fixed =[]

for entry in oracle_pool:
    id = entry["id"]
    if id not in dup:
        oracle_pool_fixed.append(entry)
        dup.add(id)
```

```python
print(f"length of base pool : {len(oracle_pool_fixed)}")
```

```python
output_path  = "evaluation/pubmedqa_oracle_pool_fixed.json"
with open(output_path, "w", encoding="utf-8") as f:
    json.dump(oracle_pool_fixed, f, indent=2)
```
